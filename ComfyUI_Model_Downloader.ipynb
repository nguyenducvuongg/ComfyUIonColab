# ⬇️ ComfyUI Model Downloader (Google Drive)
### Trình tải mô hình AI đa luồng siêu tốc (Aria2c + HuggingFace Token + Civitai API Key)
Công cụ hỗ trợ tải trực tiếp các mô hình AI (Checkpoints, LoRA, VAE, ControlNet, Background Removal, Video Models...) tốc độ cao vào Google Drive thông qua Colab.

**Hướng dẫn:**
1. Chạy **Bước 1** để kết nối Google Drive và cài đặt aria2.
2. Điền link vào form ở **Bước 2** và bấm chạy để tải (Hỗ trợ tải đơn lẻ hoặc dán danh sách tải hàng loạt).

```python
#@title 🛠️ 1. Kết nối Google Drive & Cài đặt môi trường tải
Drive_Path = "/content/drive/MyDrive/ComfyUI_Data" #@param {type:"string"}

import os
import sys
import subprocess

REPO_URL = "https://github.com/nguyenducvuongg/ComfyUIonColab.git"
REPO_DIR = "/content/ComfyUIonColab"

# Tự động đồng bộ mã nguồn ComfyColab từ GitHub
if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=False)
else:
    subprocess.run(["git", "-C", REPO_DIR, "pull", "-q"], check=False)

if REPO_DIR not in sys.path:
    sys.path.insert(0, REPO_DIR)

from comfycolab import setup_downloader_env
setup_downloader_env(Drive_Path)
```

```python
#@title 📥 2. Trình tải File thông minh (Tải Đơn Lẻ & Hàng Loạt)

Download_Mode = "Single (Tải Đơn Lẻ)" #@param ["Single (Tải Đơn Lẻ)", "Batch (Tải Hàng Loạt)"]
Skip_Existing = True #@param {type:"boolean"}

#@markdown ---
#@markdown ### ⚡ Cấu hình Tải Đơn Lẻ
Model_URL = "" #@param {type:"string"}
Model_Type = "checkpoints" #@param ["checkpoints", "loras", "vae", "clip", "clip_vision", "controlnet", "diffusion_models", "unet", "upscale_models", "ipadapter", "embeddings", "style_models", "animatediff_models", "audio_encoders", "model_patches", "text_encoders", "detection", "latent_upscale_models", "videodepthanything", "frame_interpolation", "background_removal", "SEEDVR2", "flux", "wan", "hunyuan_video"]
Custom_Folder_Name = "" #@param {type:"string"}
Custom_File_Name = "" #@param {type:"string"}

#@markdown ---
#@markdown ### 🔑 Xác thực Token (Tự động lưu vào Drive)
Civitai_API_Key = "" #@param {type:"string"}
HF_Token = "" #@param {type:"string"}

from comfycolab import run_downloader_ui

try:
    base_dir = Drive_Path
except NameError:
    base_dir = "/content/drive/MyDrive/ComfyUI_Data"

run_downloader_ui(
    download_mode=Download_Mode,
    skip_existing=Skip_Existing,
    model_url=Model_URL,
    model_type=Model_Type,
    custom_folder_name=Custom_Folder_Name,
    custom_file_name=Custom_File_Name,
    civitai_api_key=Civitai_API_Key,
    hf_token=HF_Token,
    drive_path=base_dir
)
```
